```python
from openpyxl import load_workbook, Workbook
import pandas as pd

path = r"C:\Users\user\Desktop\Connecters\new.xlsx"



def filter_data_by_fromposte(file_path):
    df = pd.read_excel(file_path, sheet_name=0)
    unique_postes = df['FromPoste'].unique()
    filtered_data = {poste: df[df['FromPoste'] == poste] for poste in unique_postes}
    return filtered_data

def write_data_to_sheets(filtered_data, output_file_path):
    """Write the filtered data to new sheets in a new Excel file."""
    with pd.ExcelWriter(output_file_path, engine='openpyxl') as writer:
        for poste, data in filtered_data.items():
            data.to_excel(writer, sheet_name=str(poste), index=False)
```

```python
filtered_data = filter_data_by_fromposte(path)
write_data_to_sheets(filtered_data, path)
```

```python
for sheet in wb.sheets:
    sheet_name = sheet.name
    print(f"Traitement de la feuille : {sheet_name}")

    # Filtrer les données pour le poste correspondant
    filtered_df = df[df['FromPoste'] == sheet_name]

    if filtered_df.empty:
        print(f"Aucune donnée trouvée pour la feuille {sheet_name}.")
        continue

    # Créer un dictionnaire des couleurs à appliquer
    shape_colors = dict(zip(filtered_df['FromCav'], filtered_df['Original\nColor']))

    # Appliquer les couleurs aux formes dans la feuille
    for shape in sheet.shapes:
        if shape.name in shape_colors:
            shape.api.Fill.ForeColor.RGB = xw.utils.rgb_to_int(shape_colors[shape.name])
            print(f"Forme '{shape.name}' colorée avec {shape_colors[shape.name]}.")

# Sauvegarder les modifications
wb.save(path)
print("Les modifications ont été enregistrées.")
```

```python
import xlwings as xw

# Dictionnaire des couleurs en hexadécimal
color_dict = {
    'Y': (255, 255, 0),   # Yellow
    'G': (0, 255, 0),     # Green
    'L': (0, 0, 255),     # Blue
    'R': (255, 0, 0),     # Red
    'W': (255, 255, 255), # White
    'LG': (128, 128, 128),# Gray (Light Gray)
    'O': (255, 165, 0),   # Orange
    'BR': (128, 71, 28),  # Brown
    'V': (128, 0, 128),   # Violet
    'GY': (112, 112, 112),# Gray
    'B': (0, 0, 0),       # Black (Noir)
    'P': (255, 192, 203), # Pink
    'C': (0, 255, 255),   # Cyan
    'D': (255, 255, 240), # Ivory
    'SI': (80, 80, 80),   # Silver
}

def apply_color(symbol, cell):
    if '/' in str(symbol):
        primary, secondary = symbol.split('/')
        # primary_rgb = tuple(int(primary[i:i+2], 16) for i in (0, 2, 4))
        primary_rgb = color_dict[primary]
        secondary_rgb = color_dict[secondary]
        cell.color = primary_rgb#int(primary_rgb, 16)
        
        cell.api.Borders(5).LineStyle = 1  
        cell.api.Borders(5).Weight = 3 
        cell.api.Borders(5).Color = secondary_rgb#int(secondary_rgb, 16)

    elif symbol in color_dict:
        # Appliquer une couleur de fond simple
        # rgb = tuple(int(color_dict[symbol][i:i+2], 16) for i in (0, 2, 4))
        # cell.color = rgb
        rgb = color_dict[symbol]
        print(f"Applying color {rgb} to cell {cell.address}")
        cell.color = rgb#int(rgb, 16)
    else:
        cell.value = symbol

# Exemple d'utilisation
app = xw.App(visible=False)

# Chemin vers le fichier Excel
path = r'C:\Users\user\Desktop\Connecters\color.xlsx'

# Ouvrir le fichier dans cette instance visible
wb = app.books.open(path)

sheet = wb.sheets[0]  # Remplacez par le nom ou l'index de votre feuille

# Appliquer la couleur et le trait diagonal à une cellule
cell = sheet.range("A2")
apply_color("R/Y", cell)  
# Sauvegarder les modifications
wb.save(path)
print("Couleur et diagonale appliquées à la cellule.")
```

```python
import xlwings as xw
from PIL import Image, ImageDraw

# Dictionnaire des couleurs et des chemins d'images correspondants
color_to_image = {
    '1': r'C:\Users\user\Desktop\blue.png',
    '2': r'C:\Users\user\Desktop\red.png',
    '6': r'C:\Users\user\Desktop\yellow.png',
    '3': r'C:\Users\user\Desktop\green.png',
    '4': r'C:\Users\user\Desktop\black.png'
}

def create_color_images():
    """Créer des images pour chaque couleur si elles n'existent pas."""
    color_map = {
        'blue': '0000FF',
        'red': 'FF0000',
        'yellow': 'FFFF00',
        'green': '00FF00',
        'black': '000000'
    }
    for name, hex_color in color_map.items():
        img = Image.new('RGB', (50, 50), f'#{hex_color}')
        print(img)
        draw = ImageDraw.Draw(img)
       
        img.save(f'C:\Users\user\Desktop\{name}.png')

def insert_images_in_excel(file_path, sheet_name, color_col, target_col):
    """Insérer des images dans Excel en fonction des couleurs."""
    wb = xw.Book(file_path)
    ws = wb.sheets[sheet_name]

    # Parcourir les lignes pour insérer les images
    for row in range(16, ws.cells.last_cell.row + 1):  # Commence à la ligne 16
        color_value = ws.range((row, color_col)).value  # Colonne des couleurs
        if color_value and str(color_value) in color_to_image:
            image_path = color_to_image[str(color_value)]
            cell = ws.range((row, target_col))  # Colonne cible pour l'image
            ws.pictures.add(image_path, left=cell.left, top=cell.top, width=30, height=30)

    wb.save()
    wb.close()

# Créer les images si nécessaire
create_color_images()

# Insérer les images dans le fichier Excel
file_path = r"C:\Users\user\Desktop\Copy of c2.xlsx"
sheet_name = 'Sheet1'  # Remplacez par le nom de votre feuille
color_col = 2  # Colonne des couleurs (par exemple, "Color")
target_col = 3  # Colonne cible pour insérer les images
insert_images_in_excel(file_path, sheet_name, color_col, target_col)
```

```python
import xlwings as xw

# Chemin vers le fichier Excel
path = r"C:\Users\user\Desktop\Copy of c2.xlsx"

# Ouvrir le fichier Excel
app = xw.App(visible=False)
wb = app.books.open(path)
sheet = wb.sheets['Sheet1']

# Source et cible
source_cell = sheet.range('B16')
target_cell = sheet.range('E16')

# Copier la valeur
target_cell.value = source_cell.api.Interior.Color

# Copier la couleur de fond
target_cell.api.Interior.Color = source_cell.api.Interior.Color

# Copier les bordures
for i in range(1, 7):  # Les bordures sont indexées de 1 à 6
    target_cell.api.Borders(i).LineStyle = source_cell.api.Borders(i).LineStyle
    target_cell.api.Borders(i).Weight = source_cell.api.Borders(i).Weight
    target_cell.api.Borders(i).Color = source_cell.api.Borders(i).Color

# Sauvegarder et fermer
wb.save(path)
# wb.close()
# app.quit()

print("La cellule B16 a été copiée vers E16 avec les bordures.")
```
