# Excel Sheet Screenshot Utility

This notebook provides utilities to capture screenshots of Excel sheets and insert them as images into other sheets within the same workbook.

## Import Required Libraries

We need to import several libraries to handle Excel automation and image processing:
- xlwings for Excel interaction
- PIL (Pillow) for image processing
- Other utility libraries

```python
# Import required libraries
import xlwings as xw
from PIL import Image, ImageGrab
import os
import io
import time
import tempfile
import win32gui
import win32ui
import win32con
import win32api
from ctypes import windll
```

## Connect to Excel Workbook

First, we need to establish a connection to the Excel workbook and access its sheets.

```python
def connect_to_workbook(path):
    """
    Connect to an Excel workbook and return the workbook object.
    
    Args:
        path (str): The path to the Excel workbook
        
    Returns:
        xw.Book: The Excel workbook object
    """
    try:
        # Try to connect to an already open workbook
        wb = xw.books[os.path.basename(path)]
        print(f"Connected to already open workbook: {path}")
    except:
        # Open the workbook if it's not already open
        wb = xw.Book(path)
        print(f"Opened workbook: {path}")
    
    return wb

# Example usage
# wb = connect_to_workbook(r"C:\Users\user\Desktop\example.xlsx")
# sheets = wb.sheets
# print(f"Available sheets: {[sheet.name for sheet in sheets]}")
```

## Capture Screenshot of Excel Sheet

Create a function that captures screenshots of Excel sheets using Windows API for high-quality captures.

```python
def capture_excel_sheet(sheet, visible_only=True):
    """
    Capture a screenshot of an Excel sheet
    
    Args:
        sheet (xw.Sheet): The Excel sheet to capture
        visible_only (bool): Whether to capture only the visible part of the sheet
        
    Returns:
        PIL.Image: The captured image
    """
    # Activate the sheet and ensure Excel window is visible
    sheet.activate()
    sheet.api.Parent.Application.Visible = True
    app = sheet.api.Parent.Application
    
    # Give Excel time to update the display
    time.sleep(0.5)
    
    if visible_only:
        # Get the handle of the Excel window
        hwnd = win32gui.FindWindow("XLMAIN", None)
        
        # Get window dimensions
        left, top, right, bottom = win32gui.GetClientRect(hwnd)
        width = right - left
        height = bottom - top
        
        # Create device context
        hwndDC = win32gui.GetWindowDC(hwnd)
        mfcDC = win32ui.CreateDCFromHandle(hwndDC)
        saveDC = mfcDC.CreateCompatibleDC()
        
        # Create bitmap
        saveBitMap = win32ui.CreateBitmap()
        saveBitMap.CreateCompatibleBitmap(mfcDC, width, height)
        saveDC.SelectObject(saveBitMap)
        
        # Copy screen to bitmap
        result = windll.user32.PrintWindow(hwnd, saveDC.GetSafeHdc(), 0)
        
        # Convert to PIL Image
        bmpinfo = saveBitMap.GetInfo()
        bmpstr = saveBitMap.GetBitmapBits(True)
        img = Image.frombuffer(
            'RGB',
            (bmpinfo['bmWidth'], bmpinfo['bmHeight']),
            bmpstr, 'raw', 'BGRX', 0, 1)
        
        # Clean up
        win32gui.DeleteObject(saveBitMap.GetHandle())
        saveDC.DeleteDC()
        mfcDC.DeleteDC()
        win32gui.ReleaseDC(hwnd, hwndDC)
        
        # Crop to remove Excel UI
        # These values may need adjustment based on Excel version and display settings
        ribbon_height = 130  # Approximate height of ribbon
        formula_bar_height = 20  # Approximate height of formula bar
        statusbar_height = 25  # Approximate height of status bar
        
        img = img.crop((0, ribbon_height + formula_bar_height, 
                        width, height - statusbar_height))
        
        return img
    else:
        # Alternative method using clipboard for full sheet (may be less reliable)
        app.ScreenUpdating = True
        
        # Select all cells with content
        sheet.api.UsedRange.Select()
        
        # Copy to clipboard
        app.Selection.CopyPicture(Format=2)  # 2 = xlBitmap
        
        # Get image from clipboard
        img = ImageGrab.grabclipboard()
        
        # Restore selection
        app.CutCopyMode = False
        
        return img
```

## Insert Screenshot into Another Sheet

Now we create a function to insert the captured screenshot into another sheet in the workbook.

```python
def insert_image_into_sheet(sheet, image, cell_address="A1", height=None, width=None):
    """
    Insert an image into an Excel sheet at the specified cell
    
    Args:
        sheet (xw.Sheet): The Excel sheet to insert the image into
        image (PIL.Image): The image to insert
        cell_address (str): The cell address to insert the image at
        height (float, optional): The height of the image in points
        width (float, optional): The width of the image in points
        
    Returns:
        xw.Shape: The inserted image shape
    """
    # Save image to temporary file
    with tempfile.NamedTemporaryFile(suffix='.png', delete=False) as temp_file:
        temp_path = temp_file.name
        image.save(temp_path, 'PNG')
    
    # Insert image into sheet
    left = sheet.range(cell_address).left
    top = sheet.range(cell_address).top
    
    # Add picture and get the shape object
    picture = sheet.pictures.add(temp_path, left=left, top=top, update=True)
    
    # Resize if dimensions provided
    if height is not None and width is not None:
        picture.height = height
        picture.width = width
    elif height is not None:
        # Maintain aspect ratio if only height provided
        aspect_ratio = picture.width / picture.height
        picture.height = height
        picture.width = height * aspect_ratio
    elif width is not None:
        # Maintain aspect ratio if only width provided
        aspect_ratio = picture.height / picture.width
        picture.width = width
        picture.height = width * aspect_ratio
    
    # Clean up temporary file
    os.unlink(temp_path)
    
    return picture
```

## Complete Workflow Example

Now let's demonstrate the entire workflow with a practical example: capturing a specific sheet and inserting it into another sheet in the same workbook.

```python
def capture_and_insert_sheet(workbook_path, source_sheet_name, target_sheet_name, cell_address="A1", width=None):
    """
    Capture a screenshot of a source sheet and insert it into a target sheet
    
    Args:
        workbook_path (str): Path to the Excel workbook
        source_sheet_name (str): Name of the sheet to capture
        target_sheet_name (str): Name of the sheet to insert the image into
        cell_address (str): Cell address for image placement
        width (float, optional): Width to resize the image to (maintains aspect ratio)
        
    Returns:
        xw.Shape: The inserted image shape
    """
    # Connect to workbook
    wb = connect_to_workbook(workbook_path)
    
    # Get source and target sheets
    try:
        source_sheet = wb.sheets[source_sheet_name]
    except:
        print(f"Error: Source sheet '{source_sheet_name}' not found")
        return None
    
    try:
        target_sheet = wb.sheets[target_sheet_name]
    except:
        print(f"Creating new target sheet '{target_sheet_name}'")
        target_sheet = wb.sheets.add(target_sheet_name)
    
    # Capture screenshot of source sheet
    print(f"Capturing screenshot of sheet '{source_sheet_name}'...")
    img = capture_excel_sheet(source_sheet)
    
    if img is None:
        print("Error: Failed to capture screenshot")
        return None
    
    # Insert image into target sheet
    print(f"Inserting screenshot into sheet '{target_sheet_name}' at {cell_address}...")
    picture = insert_image_into_sheet(target_sheet, img, cell_address, width=width)
    
    print("Screenshot successfully captured and inserted!")
    return picture

# Example usage
# capture_and_insert_sheet(
#     r"C:\Users\user\Desktop\example.xlsx", 
#     "DataSheet", 
#     "Summary", 
#     "B5", 
#     width=400
# )
```

## Additional Utility Functions

Here are some additional utility functions that might be useful when working with Excel screenshots.

```python
def capture_multiple_sheets(workbook_path, sheet_names, target_sheet_name):
    """
    Capture screenshots of multiple sheets and insert them into a target sheet
    
    Args:
        workbook_path (str): Path to the Excel workbook
        sheet_names (list): List of sheet names to capture
        target_sheet_name (str): Name of the sheet to insert the images into
    """
    wb = connect_to_workbook(workbook_path)
    
    # Create target sheet if it doesn't exist
    try:
        target_sheet = wb.sheets[target_sheet_name]
    except:
        target_sheet = wb.sheets.add(target_sheet_name)
    
    # Set starting position
    current_row = 1
    
    for sheet_name in sheet_names:
        # Add sheet name as header
        target_sheet.range(f"A{current_row}").value = f"Sheet: {sheet_name}"
        target_sheet.range(f"A{current_row}").api.Font.Bold = True
        current_row += 1
        
        # Capture and insert screenshot
        try:
            source_sheet = wb.sheets[sheet_name]
            img = capture_excel_sheet(source_sheet)
            
            if img is not None:
                picture = insert_image_into_sheet(
                    target_sheet, 
                    img, 
                    f"A{current_row}", 
                    width=600
                )
                
                # Adjust row position for next image
                # Add height of image plus some padding
                current_row += int(picture.height / 15) + 2  # Approximate row height
            else:
                target_sheet.range(f"A{current_row}").value = "Error capturing screenshot"
                current_row += 1
        except Exception as e:
            target_sheet.range(f"A{current_row}").value = f"Error: {str(e)}"
            current_row += 1
        
        # Add spacing between sheets
        current_row += 2
    
    # Save workbook
    wb.save()
    print(f"All sheet screenshots captured and inserted into '{target_sheet_name}'")

# Example usage
# capture_multiple_sheets(
#     r"C:\Users\user\Desktop\example.xlsx", 
#     ["Sheet1", "Sheet2", "Sheet3"], 
#     "Screenshots Summary"
# )
```
