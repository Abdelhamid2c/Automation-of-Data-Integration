```python
from openpyxl import load_workbook, Workbook
import pandas as pd

path = r"C:\Users\user\Desktop\Connecters\new.xlsx"



def filter_data_by_fromposte(file_path):
    df = pd.read_excel(file_path, sheet_name=0)
    unique_postes = df['FromPoste'].unique()
    filtered_data = {poste: df[df['FromPoste'] == poste] for poste in unique_postes}
    return filtered_data

def write_data_to_sheets(filtered_data, output_file_path):
    """Write the filtered data to new sheets in a new Excel file."""
    with pd.ExcelWriter(output_file_path, engine='openpyxl') as writer:
        for poste, data in filtered_data.items():
            data.to_excel(writer, sheet_name=str(poste), index=False)
```

```python
filtered_data = filter_data_by_fromposte(path)
write_data_to_sheets(filtered_data, path)
```

```python
for sheet in wb.sheets:
    sheet_name = sheet.name
    print(f"Traitement de la feuille : {sheet_name}")

    # Filtrer les données pour le poste correspondant
    filtered_df = df[df['FromPoste'] == sheet_name]

    if filtered_df.empty:
        print(f"Aucune donnée trouvée pour la feuille {sheet_name}.")
        continue

    # Créer un dictionnaire des couleurs à appliquer
    shape_colors = dict(zip(filtered_df['FromCav'], filtered_df['Original\nColor']))

    # Appliquer les couleurs aux formes dans la feuille
    for shape in sheet.shapes:
        if shape.name in shape_colors:
            shape.api.Fill.ForeColor.RGB = xw.utils.rgb_to_int(shape_colors[shape.name])
            print(f"Forme '{shape.name}' colorée avec {shape_colors[shape.name]}.")

# Sauvegarder les modifications
wb.save(path)
print("Les modifications ont été enregistrées.")
```

```python
import xlwings as xw

color_dict = {
    'Y': (255, 255, 0),   # Yellow
    'G': (0, 255, 0),     # Green
    'L': (0, 0, 255),     # Blue
    'R': (255, 0, 0),     # Red
    'W': (255, 255, 255), # White
    'LG': (128, 128, 128),# Light Gray
    'O': (255, 165, 0),   # Orange
    'BR': (128, 71, 28),  # Brown
    'V': (128, 0, 128),   # Violet
    'GY': (112, 112, 112),# Gray
    'B': (0, 0, 0),       # Black
    'P': (255, 192, 203), # Pink
    'C': (0, 255, 255),   # Cyan
    'D': (255, 255, 240), # Ivory
    'SI': (80, 80, 80),   # Silver
}

def apply_color(symbol, cell):
    if '/' in str(symbol):
        primary, secondary = symbol.split('/')
        primary_rgb = color_dict.get(primary)
        secondary_rgb = color_dict.get(secondary)

        if primary_rgb and secondary_rgb:
            cell.color = primary_rgb

            try:
                diagonal = cell.api.Borders(5)  
                diagonal.LineStyle = 1 
                diagonal.Weight = 3     
                # Convertir RGB en BGR (nécessaire pour COM)
                diagonal.Color = secondary_rgb[2] << 16 | secondary_rgb[1] << 8 | secondary_rgb[0]
            except Exception as e:
                print(f"Erreur lors de l'application de la bordure : {e}")

    elif symbol in color_dict:
        rgb = color_dict[symbol]
        print(f"Applying color {rgb} to cell {cell.address}")
        cell.color = rgb
    else:
        cell.value = symbol

path = r"C:\Users\user\Desktop\Connecters\color.xlsx"
wb = xw.Book(path)
sheet = wb.sheets[0]

cell = sheet.range("B2")
apply_color("R/Y", cell)
cell = sheet.range("B3")
apply_color("Y/W", cell)
cell = sheet.range("B4")
apply_color("G/L", cell)
cell = sheet.range("B5")
apply_color("B", cell)

wb.save(path)
print("Couleur et diagonale appliquées à la cellule.")
```

```python


def color_cavity(sheet_shapes, sheet_data):
    data_range = sheet_data.used_range

    
    try:
        color_col_index = header_row.index("color")
        shape_name_col_index = 0 
        
        for row_index in range(1, data_range.rows.count):
            row_values = data_range.rows[row_index].value
            shape_name = row_values[shape_name_col_index]
            
            color_cell = data_range.cells(row_index + 1, color_col_index + 1)
            font_color = color_cell.api.Font.Color
            
            if font_color is not None:
                shape_colors[shape_name] = font_color
                print(f"Couleur de police récupérée pour '{shape_name}': {font_color}")
    
    except ValueError:
        print("Colonne 'color' non trouvée dans les données.")
        return
    
    for shape in sheet_shapes.shapes:
        if shape.name in shape_colors:
            shape.api.Fill.ForeColor.RGB = shape_colors[shape.name]
            print(f"Forme '{shape.name}' colorée avec la couleur de police {shape_colors[shape.name]}.")

    wb.save(path)


color_cavity(sheet_shapes, sheet_data)
```
